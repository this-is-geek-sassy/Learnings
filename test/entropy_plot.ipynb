```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# n-ary entropy: one symbol has probability p, the remaining (n-1) symbols share (1-p) equally.
n = 10000
p = np.linspace(1e-6, 1-1e-6, 500)
others = (1 - p) / (n - 1)

# Entropy formula
H = -(p * np.log2(p) + (n - 1) * others * np.log2(others))


plt.figure(figsize=(6,4))
plt.plot(p, H)
plt.xlabel("p (probability of one symbol)")
plt.ylabel("Entropy H(p)")
plt.title(f"{n}-ary Entropy (one symbol varies, others equal)")
plt.grid(True)
plt.tight_layout()
plt.show()
```
